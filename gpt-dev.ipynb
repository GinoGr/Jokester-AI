```python
#Adding the dataset we will use for training.
import kagglehub

# Download latest version
path = kagglehub.dataset_download("abhinavmoudgil95/short-jokes")

print("Path to dataset files:", path)
```

```python
with open('shortjokes.csv', 'r', encoding = 'utf-8') as f:
    text = f.read()
```

```python
print('Length of text:', len(text))
```

```python
# Show first 1000 chars
print(text[:1000])
```

```python
#Get all unique chars
chars = sorted(list(set(text)))
vocab_size = len(chars)
print(''.join(chars))
print(vocab_size)
```

```python
#Tokenize chars
stoi = { ch:i for i,ch in enumerate(chars) }
itos = { i:ch for i,ch in enumerate(chars) }
encode = lambda s: [stoi[ch] for ch in s] #enccoder: take a stirng, output a lsit of ints
decode = lambda l: ''.join(  [itos[i] for i in l]) #decoder: takes lists of intergers outputs a string

print(encode('Hello world'))
print(decode(encode('Hello world')))
```
